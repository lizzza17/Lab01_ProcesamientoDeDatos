#### MENOR DE DOS PARES: Escriba una función que devuelva el menor de dos números dados *si* ambos números son pares, pero devuelva el mayor si uno o ambos números son impares
    lesser_of_two_evens(2,4) --> 2
    lesser_of_two_evens(2,5) --> 5

```python
def lesser_of_two_evens(a,b):
    pass
    #Colocar aqui la respuesta
    if a % 2 == 0 and b % 2 == 0:#verifica si ambos números son pares
        return min(a, b)
    else:
        return max(a, b)#verifica si alguno de los dos números es impar y devuelve el mayor de los dos
        
```

```python
# Revisión
lesser_of_two_evens(2,4)
```

```python
# Revisión
lesser_of_two_evens(2,5)
```

#### GALLETAS DE ANIMALES: Escribir una función que toma una cadena de dos palabras y devuelve True si ambas palabras comienzan con la misma letra
    animal_crackers('Ciudad Calor') --> True
    animal_crackers('Lugar Frio') --> False

```python
def animal_crackers(text):
  pass
  if text.split()[0][0].lower() == text.split()[1][0].lower(): #verifica si la primera letra de la primera palabra es igual a la primera letra de la segunda palabra, ignorando mayúsculas y minúsculas
    return True
  else:
    return False
```

```python
# Revisión
animal_crackers('Ciudad Calor')
```

```python
# Revisión
animal_crackers('Lugar Frio')
```

#### HACE VEINTE: dados dos enteros, devuelve True si la suma de los enteros es 20 o si uno de los enteros es 20. De lo contrario, devuelve False

    makes_twenty(20,10) --> True
    makes_twenty(12,8) --> True
    makes_twenty(2,3) --> False

```python
def makes_twenty(n1,n2):
  pass
  if n1 + n2 == 20 or n1 == 20 or n2 == 20: #verifica si la suma de los dos números es igual a 20 o si alguno de los dos números es igual a 20
    return True
  else:
    return False
```

```python
# Revisión
makes_twenty(20,10)
```

```python
# Revisión
makes_twenty(2,3)
```

```python
makes_twenty(12,8)
```

#### MAYÚSCULAS: Escribir una función que haga mayúscula la primera y cuarta letra de una cadena de texto
     
    old_macdonald('macdonald') --> MacDonald
    
Note: `'macdonald'.capitalize()` returns `'Macdonald'`

```python
def old_macdonald(name):
    pass
    #Colocar aqui la respuesta
    primera_parte = name[:3].capitalize()#con los slices se toma la primera parte del nombre y se capitaliza
    segunda_parte = name[3:].capitalize()  #igual que la primera parte, pero se toma desde el cuarto caracter hasta el final y se capitaliza
    
    return primera_parte + segunda_parte #se unen
```

```python
# Check
old_macdonald('macdonald')
```

#### REVERSA: Dada una oración, retornar la misma oración pero en reversa

    master_yoda('I am home') --> 'home am I'
    master_yoda('We are ready') --> 'ready are We'
    
Note: Revisar método .join()

    >>> "--".join(['a','b','c'])
    >>> 'a--b--c'

Resultado de .join():

    >>> " ".join(['Hello','world'])
    >>> "Hello world"

```python
def master_yoda(text):
    pass
    palabras = text.split() #separa el texto en palabras usando el espacio como delimitador
    palabras_invertidas = palabras[::-1] #invierte la lista de palabras

    return " ".join(palabras_invertidas)
```

```python
# Revisión
master_yoda('I am home')
```

```python
# Revisión
master_yoda('We are ready')
```

#### PROBLEMA 33: 

Dada una lista de enteros, retornar True si el arreglo tinene un  3  seguido por un  3 .

    has_33([1, 3, 3]) → True
    has_33([1, 3, 1, 3]) → False
    has_33([3, 1, 3]) → False

```python
def has_33(nums):
    pass
    for i in range(len(nums) - 1): # Recorremos la lista hasta el penúltimo elemento
        if nums[i] == 3 and nums[i + 1] == 3: # Verificamos si el elemento actual y el siguiente son ambos 3
            return True
            
    return False
```

```python
# Revisión
has_33([1, 3, 3])
```

```python
# Revisión
has_33([1, 3, 1, 3])
```

```python
# Revisión
has_33([3, 1, 3])
```

#### REPLICADOR: Dada una cadena, devuelve una cadena donde por cada carácter en el original hay tres caracteres REPETIDOS
    paper_doll('Hello') --> 'HHHeeellllllooo'
    paper_doll('Mississippi') --> 'MMMiiissssssiiippppppiii'

```python
def paper_doll(text):
    pass
    resultado = ''
    for char in text:
        resultado += char * 3
    return resultado
```

```python
# Check
paper_doll('Hello')
```

```python
# Check
paper_doll('Mississippi')
```

#### BLACKJACK: Dados tres enteros entre 1 y 11, si su suma es menor o igual a 21, devuelve su suma. Si su suma excede 21 *y* hay un once, reduce la suma total en 10. Finalmente, si la suma (incluso después del ajuste) excede 21, devuelve 'BUST'
    blackjack(5,6,7) --> 18
    blackjack(9,9,9) --> 'BUST'
    blackjack(9,9,11) --> 19

```python
def blackjack(a,b,c):
    pass
    suma_total = a + b + c #suma de los tres números
    if suma_total > 21 and 11 in (a, b, c):#condición para restar 10 si hay un 11 y la suma es mayor a 21, partiendo la baraja
        suma_total -= 10
    if suma_total <= 21: #condición para devolver la suma total si no se pasa de 21
        return suma_total
    else:
        return 'BUST'  #si excede 21 y no hay un 11, devuelve 'BUST'
```

```python
# Revisión
blackjack(5,6,7)
```

```python
# Revisión
blackjack(9,9,9)
```

```python
# Revisión
blackjack(9,9,11)
```
